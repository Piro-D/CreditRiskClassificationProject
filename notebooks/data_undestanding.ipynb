```python
import pandas as pd
```

```python
df = pd.read_csv(
    "../Dataset/raw/german.data",
    sep=" ",
    header=None
)

print(df.shape)
```

```python
print(df.info())
```

```python
# Data is encoded into Ax format, which require to be decoded with information from the website/description file
# Headers are not provided in the dataset, so we need to add them manually

columns = [
    "checking_status", # Qualitative
    "duration", # in months
    "credit_history", # Qualitative
    "purpose", # Qualitative
    "credit_amount", # Numerical
    "savings_status", # Qualitative
    "employment", # Qualitative
    "installment_rate", # Numerical
    "personal_status_sex", # Qualitative
    "other_debtors", # Qualitative
    "residence_duration", # in years
    "property", # Qualitative
    "age", # Numerical
    "other_installment_plans", # Qualitative
    "housing", # Qualitative
    "existing_credits", # Numerical
    "job", # Qualitative
    "num_dependents", # Numerical
    "telephone", # Qualitative
    "foreign_worker", # Qualitative
    "credit_risk" # numerical (1 = good, 2 = bad)
]

# The cost matrix is
# |           |Predict Good |Predict Bad |
# |-----------|-------------|------------|
# | ActualGood|     0       |     1      |
# | ActualBad |     5       |     0      |
# It is worse to class a customer as good when they are bad (5), 
# than it is to class a customer as bad when they are good (1).


df.columns = columns
print(df.info())
```
